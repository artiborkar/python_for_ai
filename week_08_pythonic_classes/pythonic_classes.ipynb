### concept-2--from_dict

```python
class Person:
    def __init__(self,name,age ):
        self.name=name
        self.age=age

    @classmethod
    def from_string(cls,data):
        name,age=data.split(",")
        return cls(name,int(age))

    @classmethod
    def from_dict(cls,data):
        name,age=data["name"],data["age"]
        return cls(name,age)
```

```python
p1=Person("arti",21)
```

```python
p2=Person.from_string("arti,21")
```

```python
p2.name,p2.age
```

```python
p3=Person.from_dict({"name":"Arti","age":21})
```

## combinations

```python
class Person:
    def __init__(self,name,age ):
        self.name=name
        self.age=age

    @classmethod
    def from_string(cls,data):
        name,age=data.split(",")
        return cls(name,int(age))

    @classmethod
    def from_dict(cls,data):
        name,age=data["name"],data["age"]
        return cls(name,age)

    def __str__(self):
        return  f"Person : {self.name},{self.age} year old"

    def __repr__(self):
        return f"Person(name={self.name},age={self.age})"
```

## CLASS 49     property

#### concept--1==computed property(apne aap calculate)

```python
class Circle:
    def __init__(self,radius):
        self.radius=radius
        
    @property
    def area(self):
        return 3.14*self.radius**2
        
        
```

```python
circle =Circle(3)
```

```python
circle.area
```

```python
circle.radius=5
```

```python
circle.area
```

```python
class Temperature:
    def __init__(self,celsius=0):
        self._celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self,value):
        if value < -273.15:
            raise ValueError("Below Absolute Zeroo")

        self._celsius=value
    
```

```python
temp=Temperature()
```

```python
temp.celsius
```

```python
temp.celsius=4
```

```python
temp.celsius
```

```python
print(temp.celsius)
```

```python
temp.celsius=-275
```

## CLASS 50   COMPOSITION

```python
class Engine:
    def __init__(self,horsepower):
        self.horsepower=horsepower

    def start(self):
        return f"{self.horsepower} HP Engine started"

class Car:
    def __init__(self,brand,horsepower):
        self.brand=brand
        self.horsepower=horsepower
        self.engine=Engine(self.horsepower).start()

    def start(self):
        return self.engine

```

```python
Engine(4).start()
```

```python
Car("Tata",1500).start()
```

##  Class Inside Class

```python
class Engine:
    def __init__(self,horsepower):
        self.horsepower=horsepower

    def start(self):
        return f"{self.horsepower} HP Engine started"

class Car:

    class Features:
        def __init__(self,colour):
            self.colour = colour

    def __init__(self,brand,horsepower,colour="red"):
        self.brand=brand
        self.horsepower=horsepower
        self.engine=Engine(self.horsepower).start()
        self.colour=self.colour()

    
    def start(self):
        return self.engine

    def colour(self,colour="red"):
        return self.Features(colour).colour

```

```python
Car("Tata",1300).colour
```
