## CLASS 32 — Defaults & Keyword Arguments

#### Concept 1 — default values (optional input)

```python
def greet(user_name , greeting="Hello"):
    print(f"{greeting},{user_name}")
greet("Arti")
```

```python
greet("Navneet" , "How Are You")
```

```python
greet("Arti Borkar" , "Hiii")
```

### 👨‍🏫 ⚠️ Rule — defaults hamesha BAAD mein


```python
def greeting(name,greet="hello"):
    print(f"{greet},{name}")
greeting("Arti")
```

```python
def greeting(name,greet="hello"):
    print(f"{greet},{name}")
greeting("Hi", "Arti")
```

```python
def greeting(greet="hello",name):
    print(f"{greet},{name}")
greeting("Hi", "Arti")
```

###  Concept 2 — positional vs keyword arguments

```python
def order(item,qty=1,size="medium"):
    print(f"Your oder is {item} , Quantity is {qty}, size is {size}")
```

```python
order("pizza")
```

```python
order("pizza" , 2)
```

```python
order("pizza",2,"large")
```

```python
order("small")
```

```python
order("pizza",size="small")
```
