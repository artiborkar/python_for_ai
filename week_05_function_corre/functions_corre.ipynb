### CLASS 33 — *args / **kwargs

#### Concept 1 — *args (jitne marzi values)

```python
def my_sum(*args):
    return sum(args)

print(my_sum(10, 20, 30))
```

```python
def my_args(**key_word):

    return key_word

print(my_args(name="arti",age=20))
```

```python
def total(*numbers):
    return sum(numbers)

print(total(1, 2))           # 3
print(total(1, 2, 3, 4, 5))  # 15
print(total())   
```

```python
def show_profile(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

show_profile(name="Asha", age=17, city="Mumbai")
```

```python
def show_profile(**details):
    return details
print(show_profile(name="Asha", age=17, city="Mumbai"))
```

```python
def show_profile(**details):
    return details

profile = show_profile(name="Asha", age=17, city="Mumbai")
print(profile)
```
