### 👨‍🏫 Concept 4 — attributes vs methods

📖 Technical definition — Attribute vs method: An attribute is a piece of data stored on an object (its state). A method is a function defined inside a class that operates on the object (its behaviour). Together they make up what an object "has" and what it "does."

```python
class Info:
    def __init__(self,name,age,city):
        self.name = name
        self.age = age            ## this is attributes
        self.city = city
    def detail(self):
        print(f"hello {self.name} you are {self.age } yr from {self.city}")
```

Internal state is marked private (by convention with a leading underscore) and changed only through controlled, validated methods.

```python
class BankAccount:
    def __init__(self,balance):
        self._balance = balance

    def deposite(self,amount):
        if amount > 0:
            self._balance+=amount
        return print(f"Avaliable balance is {self._balance}")

    def withdrow(self,amount):
        if amount > 0:
            self._balance-=amount
        return print(f"Avaliable balance is {self._balance}")

    def get_balance(self):
        return self._balance
```

#### Concept 2 — _ (underscore) convention — "yeh private hai"

```python
class BankAccount:
    def __init__(self,balance:int):
        self._balance = balance

    def deposite(self,amount:int)->None:
        if amount < 0 :
            return print("Invalide Amount")
        if amount > 0:
            self._balance+=amount
        return print(f"Avaliable balance is {self._balance}")

    def withdrow(self,amount:int)->None:
        if amount < 0 :
            return print("Invalide Amount")
        if amount > 0:
            self._balance-=amount
        return print(f"Avaliable balance is {self._balance}")

    def get_balance(self)->int:
        return self._balance
```

```python
bank_obj_under = BankAccount(5000)
```

## CLASS 43 — Inheritance

 ### Concept 1 — basic inheritance

📖 Technical definition — Inheritance: Inheritance lets a child (derived) class reuse the attributes and methods of a parent (base) class. The child automatically gains the parent's behaviour and can add new members or modify inherited ones, avoiding duplicated code.

```python
class Animal:
    def __init__(self,name):
        self.name = name

    def sleep(self):
        print(f"{self.name} is sleeping")

    def eat(self):
        print(f"{self.name} is eating")
```

```python
animal_obj = Animal("dog")
```

```python
animal_obj.sleep()
```

```python
animal_obj.eat()
```

```python
class Dog(Animal):
    def bark(self):
        print(f"{self.name}is barking!")
```

```python
dog_1_obj = Dog("Tommy")
```

```python
dog_1_obj.bark()
```

```python
dog_1_obj.sleep()
```

```python
dog_1_obj.eat()
```

```python
animal_obj.bark()
```

### 👨‍🏫 Concept 3 — super() (parent ka setup use karo)

📖 Technical definition — super(): super() returns a proxy to the parent class, letting a child call the parent's methods (most commonly super().__init__(...)). This reuses the parent's setup instead of duplicating it in the child

```python
class Dog(Animal):
    def __init__(self,name,age):
        super().__init__(name)
        self.age = age

    def bark(self):
        print(f"{self.name} is barking")
    
    def age_dog(self):
        print(f"{self.age} yr old {self.name}")
```

```python
dog_1_obj=Dog("Devid",8)
```

```python
dog_1_obj.bark()
```

```python
dog_1_obj.age()
```
