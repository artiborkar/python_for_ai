###  Concept 4-File ke saath JSON (dump/load)

```python
data = {
    "name" : "Arti",
    "age" : 21 , 
    "subject" : ["math" , "science"] , 
    "is_student" : True
    
       }

with open("student_record.json" , "w") as f:
    json.dump(data,f , indent=4)
```

```python
with open("student_record.json" , "r") as f:
    data_from_python = json.load(f)
```

```python
def create_history()->None:
    with open("agent_chat_history.json" , "w") as f:
        json.dump([] , f ,indent=4)

def add_history(role:str , content :str)->None:
    with open("agent_chat_history.json" , "r") as f:
        history = json.load(f)

    history.append({"role" : role , "content" : content})

    with open("agent_chat_history.json" , "w") as f:
        json.dump(history , f ,indent=4)

def add_history()->list:
    with open("agent_chat_history.json" , "r") as f:
        print(json.load(f))
```

### 👨‍🏫 Concept 3 — Regex (patterns dhoondhna)

📖 Technical definition — Regular expression (regex): A regular expression is a compact pattern language for matching, searching, and replacing text. Python's re module uses these patterns via functions such as search (first match), findall (all matches), and sub (replace matches).

regex---re library

```python
import re
```

```python
text = '''my content : email >>indianaiproduction_help@gmail.com , 
         and my mobile no>>> 1234567890 , my secondary mo no. :9876543210'''



number = re.findall(r"\d+",text)

print(number)
```

```python
list_words = text.split(" ")
```

```python
for word in list_words:
    if "@" in word:
        print(word)
```

```python
def get_email(text):
    list_words = text.split(" ")
    for word in list_words:
        if "@" in word:
            print(word)
```

```python
get_email(text)
```

```python
numbers = re.search( r"\d+",text)
print(numbers.group())
```

```python
email = re.search(r"\w+@\w+\.\w+",text)
print(email.group())
```

```python
import logging
```

```python
logging.basicConfig(level=logging.INFO , format="%(asctime)s - %(levelname)s - %(message)s")
```
